# Security Log Analysis — Core Python

This notebook analyses a synthetic security-login dataset using only Python's standard library.


```python
import json
from pathlib import Path

DATA_FILE = Path("../data/security_logs.json")

with open(DATA_FILE, "r", encoding="utf-8") as file:
    logs = json.load(file)

logs[:3]
```

```python
total_records = len(logs)
failed_logs = []
successful_logs = []
total_attempts = 0

for log in logs:
    total_attempts += log["attempts"]

    if log["status"] == "failed":
        failed_logs.append(log)
    else:
        successful_logs.append(log)

print("Total records:", total_records)
print("Failed:", len(failed_logs))
print("Successful:", len(successful_logs))
print("Total attempts:", total_attempts)
```

```python
attempts_by_user = {}

for log in logs:
    user = log["user"]
    attempts_by_user[user] = attempts_by_user.get(user, 0) + log["attempts"]

attempts_by_user
```

```python
attempts_by_ip = {}

for log in logs:
    ip = log["ip"]
    attempts_by_ip[ip] = attempts_by_ip.get(ip, 0) + log["attempts"]

attempts_by_ip
```

```python
suspicious_logs = []

for log in logs:
    if log["status"] == "failed" and log["attempts"] >= 5:
        suspicious_logs.append(log)

suspicious_logs
```

```python
summary = {
    "total_records": total_records,
    "successful_records": len(successful_logs),
    "failed_records": len(failed_logs),
    "total_attempts": total_attempts,
    "average_attempts": round(total_attempts / total_records, 2),
    "attempts_by_user": attempts_by_user,
    "attempts_by_ip": attempts_by_ip,
    "suspicious_records": suspicious_logs,
}

summary
```

```python
OUTPUT_FILE = Path("../outputs/core_notebook_summary.json")

with open(OUTPUT_FILE, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=2)

print("Wrote:", OUTPUT_FILE)
```
