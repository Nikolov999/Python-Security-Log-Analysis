# Security Log Analysis — Pandas, NumPy and Matplotlib

This notebook repeats the same analysis using common Python data-analysis libraries.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

DATA_FILE = Path("../data/security_logs.json")
df = pd.read_json(DATA_FILE)
df
```

```python
df.info()
```

```python
summary = {
    "total_records": len(df),
    "successful_records": (df["status"] == "success").sum(),
    "failed_records": (df["status"] == "failed").sum(),
    "total_attempts": np.sum(df["attempts"].to_numpy()),
    "average_attempts": np.mean(df["attempts"].to_numpy()),
    "maximum_attempts": np.max(df["attempts"].to_numpy()),
    "median_attempts": np.median(df["attempts"].to_numpy()),
}

summary
```

```python
failed = df[df["status"] == "failed"]
suspicious = failed[failed["attempts"] >= 5]
suspicious
```

```python
attempts_by_user = (
    df.groupby("user")["attempts"]
      .sum()
      .sort_values(ascending=False)
)

attempts_by_user
```

```python
attempts_by_ip = (
    df.groupby("ip")["attempts"]
      .sum()
      .sort_values(ascending=False)
)

attempts_by_ip
```

```python
attempts_by_user.plot(kind="bar", figsize=(8, 5))
plt.title("Login Attempts by User")
plt.xlabel("User")
plt.ylabel("Total Attempts")
plt.tight_layout()
plt.show()
```

```python
OUTPUT_DIR = Path("../outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

suspicious.to_csv(OUTPUT_DIR / "suspicious_logins_from_notebook.csv", index=False)
pd.DataFrame([summary]).to_json(
    OUTPUT_DIR / "library_notebook_summary.json",
    orient="records",
    indent=2
)

print("Outputs saved.")
```
